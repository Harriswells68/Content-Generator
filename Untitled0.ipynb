```python
import subprocess
import os
import math

def trim_video_ffmpeg(video_path, output_folder="video_parts", part_duration=120):
    """
    Trim video using FFmpeg (faster than MoviePy)

    Parameters:
    - video_path: Path to the input video file
    - output_folder: Folder name where parts will be saved
    - part_duration: Duration of each part in seconds (default: 120 = 2 minutes)
    """

    # Create output folder if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
        print(f"Created folder: {output_folder}")

    # Get video duration using ffprobe
    cmd = [
        'ffprobe', '-v', 'error',
        '-show_entries', 'format=duration',
        '-of', 'default=noprint_wrappers=1:nokey=1',
        video_path
    ]
    result = subprocess.run(cmd, capture_output=True, text=True)
    total_duration = float(result.stdout.strip())

    # Calculate number of parts
    num_parts = math.ceil(total_duration / part_duration)

    print(f"Video duration: {total_duration:.2f} seconds")
    print(f"Will create {num_parts} parts of {part_duration} seconds each")

    # Get video filename without extension
    video_name = os.path.splitext(os.path.basename(video_path))[0]
    video_ext = os.path.splitext(video_path)[1]

    # Split video into parts
    for i in range(num_parts):
        start_time = i * part_duration
        part_number = i + 1

        output_filename = f"{video_name}_part_{part_number:03d}{video_ext}"
        output_path = os.path.join(output_folder, output_filename)

        print(f"Creating part {part_number}/{num_parts}...")

        # FFmpeg command to extract part
        cmd = [
            'ffmpeg', '-y',
            '-i', video_path,
            '-ss', str(start_time),
            '-t', str(part_duration),
            '-c', 'copy',  # Copy codec (fast, no re-encoding)
            output_path
        ]

        subprocess.run(cmd, capture_output=True)

    print(f"\nDone! {num_parts} parts saved in '{output_folder}' folder")
    return num_parts

```

```python
"""
crop_to_reel.py
================
Crop any landscape video to TikTok / Instagram Reel format (9:16)
using FFmpeg — the fastest and most reliable approach.

Requirements:
    pip install tqdm
    FFmpeg must be installed:
        - Windows : https://ffmpeg.org/download.html  (add to PATH)
        - macOS   : brew install ffmpeg
        - Linux   : sudo apt install ffmpeg
"""

import subprocess
import json
import shutil
import sys
import re
from pathlib import Path


# ─────────────────────────────────────────────────────────────
# HELPERS
# ─────────────────────────────────────────────────────────────

def check_ffmpeg():
    """Make sure ffmpeg + ffprobe are available on PATH."""
    for tool in ("ffmpeg", "ffprobe"):
        if shutil.which(tool) is None:
            sys.exit(
                f"❌  '{tool}' not found on PATH.\n"
                "Install FFmpeg and ensure it is added to your system PATH.\n"
                "  macOS  : brew install ffmpeg\n"
                "  Linux  : sudo apt install ffmpeg\n"
                "  Windows: https://ffmpeg.org/download.html"
            )
    print("✅  FFmpeg found.")


def get_video_info(path: Path) -> dict:
    """Return a dict with width, height, fps (float), duration (float)."""
    cmd = [
        "ffprobe", "-v", "quiet",
        "-print_format", "json",
        "-show_streams", "-show_format",
        str(path),
    ]
    result = subprocess.run(cmd, capture_output=True, text=True)
    if result.returncode != 0:
        raise RuntimeError(f"ffprobe error:\n{result.stderr}")

    data = json.loads(result.stdout)

    video_stream = next(
        (s for s in data["streams"] if s["codec_type"] == "video"), None
    )
    if video_stream is None:
        raise RuntimeError("No video stream found in the file.")

    # FPS can be stored as "30/1" or "30000/1001" — eval is safe here
    fps_raw = video_stream.get("r_frame_rate", "30/1")
    num, den = map(int, fps_raw.split("/"))
    fps = num / den if den else 30.0

    duration = float(data["format"].get("duration", 0))

    return {
        "width":    int(video_stream["width"]),
        "height":   int(video_stream["height"]),
        "fps":      fps,
        "duration": duration,
    }


def hms_to_seconds(hms: str) -> float:
    """Convert 'HH:MM:SS.xx' string from FFmpeg stderr to total seconds."""
    try:
        h, m, s = hms.split(":")
        return int(h) * 3600 + int(m) * 60 + float(s)
    except Exception:
        return 0.0


# ─────────────────────────────────────────────────────────────
# MAIN FUNCTION
# ─────────────────────────────────────────────────────────────

def crop_to_reel(
    input_path: str,
    output_path: str = None,
    position: str = "center",
    target_width: int = 1080,
    target_height: int = 1920,
    quality: str = "high",
) -> str:
    """
    Crop a landscape video to 9:16 (TikTok / Instagram Reel) format.

    Parameters
    ----------
    input_path   : Path to the source video.
    output_path  : Where to save the result (auto-named if None).
    position     : Horizontal crop focus — 'center' | 'left' | 'right'
    target_width : Output width  in pixels (default 1080).
    target_height: Output height in pixels (default 1920).
    quality      : 'high' (best, slower) | 'medium' | 'fast' (quickest).

    Returns
    -------
    str : Path to the output file.
    """
    check_ffmpeg()

    input_path = Path(input_path).resolve()
    if not input_path.exists():
        raise FileNotFoundError(f"Input not found: {input_path}")

    # ── Auto-generate output path ──────────────────────────────
    # Always output .mp4 — TikTok/Reels/Shorts require H.264 in MP4.
    # WebM, MOV, AVI inputs are all re-encoded to MP4 automatically.
    if output_path is None:
        output_path = input_path.parent / f"{input_path.stem}_reel.mp4"
    output_path = Path(output_path).with_suffix(".mp4").resolve()
    output_path.parent.mkdir(parents=True, exist_ok=True)

    # ── Read source info ────────────────────────────────────────
    info = get_video_info(input_path)
    orig_w, orig_h = info["width"], info["height"]
    fps      = info["fps"]
    duration = info["duration"]

    print(f"\n📹  Input  : {input_path.name}")
    print(f"    Size   : {orig_w} × {orig_h}")
    print(f"    FPS    : {fps:.3f}")
    print(f"    Length : {duration:.2f} s  ({duration/60:.1f} min)")

    # ── Calculate crop box ──────────────────────────────────────
    # Strategy: keep full HEIGHT, shrink WIDTH to 9:16 ratio.
    # If the video is already portrait, keep full WIDTH, crop HEIGHT.
    crop_h = orig_h
    crop_w = int(orig_h * 9 / 16)

    if crop_w > orig_w:          # already portrait-ish
        crop_w = orig_w
        crop_h = int(orig_w * 16 / 9)

    # Codecs require even dimensions
    crop_w -= crop_w % 2
    crop_h -= crop_h % 2

    # Horizontal position of the crop window
    if position == "center":
        x = (orig_w - crop_w) // 2
    elif position == "left":
        x = 0
    elif position == "right":
        x = orig_w - crop_w
    else:
        raise ValueError(f"position must be 'center', 'left', or 'right'. Got: {position!r}")

    y = (orig_h - crop_h) // 2   # always vertically centered

    print(f"\n✂️   Crop   : {crop_w} × {crop_h}  (offset x={x}, y={y})")
    print(f"    → Scale : {target_width} × {target_height}")

    # ── Quality presets ─────────────────────────────────────────
    # CRF: 0=lossless … 51=worst | 18=high quality | 23=default
    quality_map = {
        "high":   ["-crf", "18", "-preset", "slow"],
        "medium": ["-crf", "23", "-preset", "medium"],
        "fast":   ["-crf", "28", "-preset", "ultrafast"],
    }
    if quality not in quality_map:
        raise ValueError(f"quality must be 'high', 'medium', or 'fast'. Got: {quality!r}")
    q_flags = quality_map[quality]
    print(f"    Quality: {quality}  (CRF {q_flags[1]})")

    # ── Build FFmpeg command ────────────────────────────────────
    # crop=w:h:x:y  →  scale  →  lanczos (best upscale/downscale filter)
    vf = (
        f"crop={crop_w}:{crop_h}:{x}:{y},"
        f"scale={target_width}:{target_height}:flags=lanczos"
    )

    cmd = [
        "ffmpeg", "-y",                         # overwrite without asking
        "-i", str(input_path),
        "-vf", vf,
        "-c:v", "libx264",                      # H.264 — universal support
        *q_flags,
        "-c:a", "aac",                          # AAC audio — TikTok/Reels standard
        "-b:a", "192k",                         # good audio quality
        "-movflags", "+faststart",              # allows streaming before full download
        "-map_metadata", "0",                   # preserve original metadata
        str(output_path),
    ]

    # ── Run with live progress ──────────────────────────────────
    print(f"\n🚀  Encoding…  (this may take a moment)\n")

    process = subprocess.Popen(
        cmd,
        stderr=subprocess.PIPE,
        stdout=subprocess.DEVNULL,
        text=True,
        universal_newlines=True,
    )

    # FFmpeg writes progress to stderr — parse `time=HH:MM:SS.xx`
    time_pattern = re.compile(r"time=(\d{2}:\d{2}:\d{2}\.\d+)")
    last_pct = -1

    for line in process.stderr:
        match = time_pattern.search(line)
        if match and duration > 0:
            current = hms_to_seconds(match.group(1))
            pct = min(int(current / duration * 100), 100)
            if pct != last_pct:
                bar = "█" * (pct // 5) + "░" * (20 - pct // 5)
                print(f"\r    [{bar}] {pct:3d}%  {current:.1f}s / {duration:.1f}s", end="", flush=True)
                last_pct = pct

    process.wait()
    print()   # newline after progress bar

    if process.returncode != 0:
        raise RuntimeError(
            "❌  FFmpeg failed. Check the command above or run it manually to see the full error."
        )

    # ── Verify output ───────────────────────────────────────────
    out_info     = get_video_info(output_path)
    out_duration = out_info["duration"]
    size_mb      = output_path.stat().st_size / (1024 * 1024)

    print(f"\n✅  Done!")
    print(f"    Output : {output_path.name}")
    print(f"    Size   : {out_info['width']} × {out_info['height']}")
    print(f"    Length : {out_duration:.2f} s  (original: {duration:.2f} s)")
    print(f"    File   : {size_mb:.1f} MB")

    if abs(out_duration - duration) > 0.5:
        print("⚠️   Duration mismatch > 0.5 s — investigate the source file.")

    return str(output_path)


# ─────────────────────────────────────────────────────────────
# BATCH HELPER
# ─────────────────────────────────────────────────────────────

def batch_crop(
    input_folder: str,
    output_folder: str = None,
    extensions: tuple = (".mp4", ".mov", ".avi", ".mkv", ".webm"),
    **kwargs,
):
    """
    Crop every video in a folder to Reel format.

    Parameters
    ----------
    input_folder  : Folder containing source videos.
    output_folder : Where to save results (defaults to input_folder/reels/).
    extensions    : Which file types to process.
    **kwargs      : Passed directly to crop_to_reel().
    """
    input_folder  = Path(input_folder)
    output_folder = Path(output_folder) if output_folder else input_folder / "reels"
    output_folder.mkdir(parents=True, exist_ok=True)

    videos = [f for f in input_folder.iterdir() if f.suffix.lower() in extensions]
    if not videos:
        print(f"No videos found in {input_folder}")
        return

    print(f"Found {len(videos)} video(s) to process.\n")
    for i, video in enumerate(videos, 1):
        print(f"── [{i}/{len(videos)}] {video.name} ──")
        out = output_folder / f"{video.stem}_reel{video.suffix}"
        try:
            crop_to_reel(str(video), str(out), **kwargs)
        except Exception as e:
            print(f"   ⚠️  Skipped: {e}")
        print()


# ─────────────────────────────────────────────────────────────
# ENTRY POINT
# ─────────────────────────────────────────────────────────────

```

```python
import subprocess
import os

def download_youtube_video_720p(video_id: str, output_dir: str = "."):
    """
    Downloads a YouTube video at 720p resolution using yt-dlp.

    Args:
        video_id: The YouTube video ID.
        output_dir: The directory where the video will be saved.
    """
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
        print(f"Created output directory: {output_dir}")

    video_url = f"https://www.youtube.com/watch?v={video_id}"
    output_template = os.path.join(output_dir, f"{video_id}.%(ext)s")

    # Command to download the best video stream with height <= 720p and best audio
    command = [
        "yt-dlp",
        "-f", "bestvideo[height<=720]+bestaudio/best[height<=720]",
        "--output", output_template,
        video_url,
    ]

    print(f"Downloading video {video_id} to {output_dir} at 720p...")
    try:
        result = subprocess.run(command, capture_output=False, text=False, check=True)
        print("Download successful!")
        print(result.stdout)
        if result.stderr:
            print("Errors/Warnings:")
            print(result.stderr)
    except subprocess.CalledProcessError as e:
        print(f"Error downloading video {video_id}: {e}")
        print(f"Standard Output: {e.stdout}")
        print(f"Standard Error: {e.stderr}")
    except FileNotFoundError:
        print("Error: yt-dlp command not found. Make sure yt-dlp is installed and in your PATH.")

# Example usage (uncomment to test):
# download_youtube_video_720p("dQw4w9WgXcQ", output_dir="./downloaded_videos")
```
